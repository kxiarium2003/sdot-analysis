```python
import pandas as pd

df = pd.read_csv("../data/S-DoT_NATURE_2026.06.22-06.28.csv", encoding="cp949")
print(f"데이터프레임 크기: {df.shape[0]} rows, {df.shape[1]} columns")
print(df.info())
df.head(5)
```

```python
cols = [
    "모델번호",
    "시리얼",
    "측정시간",
    "지역",
    "자치구",
    "행정동"
]

df = pd.read_csv(
    "../data/S-DoT_NATURE_2026.06.22-06.28.csv",
    encoding="cp949",
    index_col=False
)

df = df.rename(columns=dict(zip(df.columns[:6], cols)))
df.head()
```

### 결측치

```python
missing = df.isnull().sum().sort_values(ascending=False)
missing_ratio = (missing / len(df) * 100).round(2)

pd.DataFrame({
    "missing": missing,
    "ratio(%)": missing_ratio
})
```

```python
sensor_feature = (
    df.groupby("시리얼")
      .agg({
          "풍속 평균(m/s)": lambda x: x.notna().any(),
          "오존 평균(ppm)": lambda x: x.notna().any(),
          "흑구온도 평균(℃)": lambda x: x.notna().any(),
          "온도 평균(℃)": lambda x: x.notna().any()
      })
)

sensor_feature.sum()
```
