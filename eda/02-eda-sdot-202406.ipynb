## 가설: 센서마다 특정항목을 측정하는 모듈이 상이할 것이다
- 시리얼별 측정항목 확인
- 일부 센서에 특정 센서 모듈이 장착되었을 것이라는 가정을 EDA를 통해 검증

### 풍속/온도/오존 평균을 한 번이라도 측정한 센서 개수 확인

```python
wind_sensor = (
    sdot_2406[sdot_2406["풍속 평균(m/s)"].notnull()]
    ["시리얼"]
    .nunique()
)

temp_sensor = (
    sdot_2406[sdot_2406["온도 평균(℃)"].notnull()]
    ["시리얼"]
    .nunique()
)

ozone_sensor = (
    sdot_2406[sdot_2406["오존 평균(ppm)"].notnull()]
    ["시리얼"]
    .nunique()
)

print("온도 측정 센서 :", temp_sensor)
print("풍속 측정 센서 :", wind_sensor)
print("오존 측정 센서 :", ozone_sensor)
```

### 각 센서에 대해 한 번이라도 측정값이 존재하는지 확인
- 측정값이 많은 온도/풍속과 가장 적은 오존 평균만 확인
- 특정 시리얼에 대해서는 특정한 측정항목이 한 번도 관측되지 않았음을 확인
- 그렇다고 해서 오존을 측정하는 센서가 없다고 단정지을 수는 없음
- 측정기간 확인을 통해 검증할 필요가 있음

```python
sensor_measure = (
    sdot_2406
    .groupby("시리얼")[[
        "온도 평균(℃)",
        "풍속 평균(m/s)",
        "오존 평균(ppm)"
    ]]
    .apply(lambda x: x.notnull().any())
)

display(sensor_measure.head(10))
```

### 측정항목이 어느 자치구에 집중되어있는지 확인
- 센서가 어디에 설치되어있는지 확인

```python
variables = [
    "온도 평균(℃)",
    "풍속 평균(m/s)",
    "오존 평균(ppm)"
]

for var in variables:

    print(f"\n===== {var} =====")

    result = (
        sdot_2406.loc[
            sdot_2406[var].notnull(),
            ["자치구","시리얼"]
        ]
        .drop_duplicates()
        .groupby("자치구")["시리얼"]
        .nunique()
        .sort_values(ascending=False)
    )

    display(result)
```

## 측정항목별로 측정시간이 상이한지 확인
- 측정데이터가 가장 적은 오존의 경우, 소음/풍속 등의 측정항목과는 달리 상시 측정하지 않았을 것이라는 추정을 검증

```python
# datetime 변환
sdot_2406["측정시간"] = pd.to_datetime(
    sdot_2406["측정시간"],
    format="%Y-%m-%d_%H:%M:%S"
)
```

### monthly

```python
import matplotlib.pyplot as plt

variables = [
    "온도 평균(℃)",
    "풍속 평균(m/s)",
    "오존 평균(ppm)"
]

colors = ["tab:red", "tab:blue", "tab:green"]

plt.figure(figsize=(14, 5))

for var, color in zip(variables, colors):

    tmp = (
        sdot_2406.loc[
            sdot_2406[var].notnull()
        ]
        .groupby(
            sdot_2406.loc[
                sdot_2406[var].notnull(),
                "측정시간"
            ].dt.date
        )
        .size()
    )

    plt.plot(
        tmp.index,
        tmp.values,
        marker="o",
        linewidth=2,
        color=color,
        label=var
    )

plt.title("일자별 측정항목")
plt.xlabel("날짜")
plt.ylabel("측정 건수")
plt.legend()
plt.grid(alpha=0.3)

plt.tight_layout()
plt.show()
```

### daily
- 측정 데이터가 가장 적은 오존도 매일 측정되었음을 확인
- day -> time으로 세분화해서 확인
- 측정 기간이 달라서 결측이 발생한 것 아닐까 -> 라는 가설 확인

```python
date = "2024-06-15"

variables = [
    "온도 평균(℃)",
    "풍속 평균(m/s)",
    "오존 평균(ppm)"
]

colors = ["tab:red", "tab:blue", "tab:green"]

plt.figure(figsize=(15,5))

for var, color in zip(variables, colors):

    tmp = (
        sdot_2406[
            (sdot_2406["측정시간"].dt.date.astype(str) == date)
            &
            (sdot_2406[var].notnull())
        ]
        .groupby("측정시간")
        .size()
    )

    plt.plot(
        tmp.index,
        tmp.values,
        marker="o",
        markersize=3,
        linewidth=2,
        color=color,
        label=var
    )

plt.legend()
plt.grid(alpha=0.3)
plt.title(f"{date} 시간별 측정 건수")
plt.tight_layout()
plt.show()
```

## 자치구별 측정패턴 확인

```python
measurement_cols = [
    "온도 평균(℃)",
    "습도 평균(%)",
    "풍속 평균(m/s)",
    "조도 평균(lux)",
    "자외선 평균(UV)",
    "소음 평균(dB)",
    "암모니아 평균(ppm)",
    "황화수소 평균(ppm)",
    "오존 평균(ppm)"
]

sensor_pattern = (
    sdot_2406
    .groupby("시리얼")[measurement_cols]
    .apply(lambda x: x.notnull().any())
)

# 패턴 확인
sensor_pattern["측정패턴"] = sensor_pattern.apply(
    lambda row: " | ".join(row.index[row]),
    axis=1
)

# 자치구 정보 붙이기
sensor_location = (
    sdot_2406[["시리얼", "자치구"]]
    .drop_duplicates()
)

sensor_pattern = (
    sensor_pattern
    .reset_index()
    .merge(sensor_location, on="시리얼")
)

# 자치구 x 패턴
pattern_by_district = (
    sensor_pattern
    .groupby(["자치구", "측정패턴"])
    .size()
    .reset_index(name="센서개수")
    .sort_values(
        ["자치구", "센서개수"],
        ascending=[True, False]
    )
)

display(pattern_by_district)
```
