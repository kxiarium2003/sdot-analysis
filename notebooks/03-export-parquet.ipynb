# measure_time 포맷 지정

```python
query_find_new_format = f"""
    SELECT measure_time, COUNT(*) as count
    FROM '{DATA_PATH}'
    WHERE TRY_CAST(REPLACE(measure_time, '.', '-') AS TIMESTAMP) IS NULL
      AND length(measure_time) > 10  -- 쓰레기 이상치 데이터 제외
    GROUP BY measure_time
    ORDER BY count DESC
    LIMIT 10
"""

print("📌 [변환 실패한 날짜 포맷 정체 확인]")
display(duckdb.sql(query_find_new_format).pl())
```

```python
query_final_time_range = f"""
    SELECT 
        MIN(TRY_CAST(REPLACE(REPLACE(measure_time, '.', '-'), '_', ' ') AS TIMESTAMP)) AS start_time,
        MAX(TRY_CAST(REPLACE(REPLACE(measure_time, '.', '-'), '_', ' ') AS TIMESTAMP)) AS end_time
    FROM '{DATA_PATH}'
"""

print("📌 [최종 데이터 측정 시간 범위]")
display(duckdb.sql(query_final_time_range).pl())
```

# view 생성
- 시계열 분석을 할 때마다 매 쿼리에 TRY_CAST와 REPLACE 로직을 길게 쓰면 쿼리 속도에 영향
- 날짜를 TIMESTAMP 타입으로 변환한 파생 컬럼을 추가한 뷰 생성

```python
# 1. 뷰(View) 생성 쿼리
query_create_view = f"""
    CREATE OR REPLACE VIEW v_sdot_clean AS
    SELECT 
        *,
        TRY_CAST(REPLACE(REPLACE(measure_time, '.', '-'), '_', ' ') AS TIMESTAMP) AS measure_timestamp
    FROM '{DATA_PATH}'
    WHERE TRY_CAST(REPLACE(REPLACE(measure_time, '.', '-'), '_', ' ') AS TIMESTAMP) IS NOT NULL
"""

# 뷰 생성 실행
duckdb.sql(query_create_view)
print("✅ 깔끔한 시계열 분석을 위한 뷰(v_sdot_clean) 생성 완료!\n")

# 2. 뷰가 잘 만들어졌는지 상위 5개 확인
query_check_view = """
    SELECT 
        measure_time AS "원본_문자열(str)", 
        measure_timestamp AS "변환된_날짜(timestamp)"
    FROM v_sdot_clean
    LIMIT 5
"""

print("📌 [뷰 데이터 미리보기]")
display(duckdb.sql(query_check_view).pl())
```

# [0] 위경도 정보 매핑 (view 업데이트)

```python
import pandas as pd
import duckdb

# 1. 엑셀 파일 로드 및 컬럼명 변경
loc_df = pd.read_excel('../metadata/서울시 도시데이터 센서(S-DoT) 환경정보 설치 위치정보.xlsx') #[cite: 1]
loc_df = loc_df.rename(columns={'모델 시리얼(*)': '시리얼'}) #[cite: 1]

# 2. 변경 전 시리얼 매핑 테이블 구축[cite: 1]
old_serial_map = {}
for _, row in loc_df.iterrows():
    if pd.notna(row['변경 전 시리얼']): #[cite: 1]
        old_serial_map[row['변경 전 시리얼']] = row['시리얼'] #[cite: 1]
    if pd.notna(row['변경 전 시리얼(데이터 상 표기)']): #[cite: 1]
        old_serial_map[row['변경 전 시리얼(데이터 상 표기)']] = row['시리얼'] #[cite: 1]

# DuckDB에 위치 데이터 프레임 등록
duckdb.register('loc_table', loc_df)

# 3. 매핑 로직을 반영한 최종 View 생성
# 원본 데이터의 sensor_id가 old_serial_map에 있으면 신규 시리얼로 변경 후 JOIN
query_create_final_view = """
    CREATE OR REPLACE VIEW v_sdot_final AS
    WITH mapped_data AS (
        SELECT 
            *,
            COALESCE(
                (SELECT 시리얼 FROM loc_table WHERE "변경 전 시리얼" = d.sensor_id OR "변경 전 시리얼(데이터 상 표기)" = d.sensor_id LIMIT 1), 
                sensor_id
            ) AS sensor_id_matched
        FROM v_sdot_clean d
    )
    SELECT 
        m.*,
        l.위도 AS lat,
        l.경도 AS lon,
        l.주소 AS address
    FROM mapped_data m
    LEFT JOIN loc_table l ON m.sensor_id_matched = l.시리얼
"""

# 뷰 생성 
duckdb.sql(query_create_final_view)
print("✅ 위경도 매핑이 완료된 최종 뷰(v_sdot_final) 생성 완료!")

# 생성된 뷰 SELECT 문으로 조회
query_check_final = "SELECT * FROM v_sdot_final LIMIT 5"
display(duckdb.sql(query_check_final).pl())
```

```python
# 1. 자치구별 고유 센서 분포 확인
query_district_check = """
    WITH sensor_districts AS (
        SELECT 
            sensor_id_matched,
            address,
            SPLIT_PART(address, ' ', 2) AS district
        FROM v_sdot_final
        GROUP BY sensor_id_matched, address
    )
    SELECT 
        district AS 자치구,
        COUNT(*) AS 센서_개수
    FROM sensor_districts
    GROUP BY district
    ORDER BY 센서_개수 DESC
"""

df_district_check = duckdb.sql(query_district_check).pl()
print("📌 [정상 매핑 확인: 자치구별 센서 개수]")
display(df_district_check)

print("-" * 50)

# 2. 매핑 이상치(결측치 또는 '구'로 끝나지 않는 주소) 추출
query_anomaly_check = """
    WITH sensor_districts AS (
        SELECT 
            sensor_id_matched,
            address,
            SPLIT_PART(address, ' ', 2) AS district
        FROM v_sdot_final
        GROUP BY sensor_id_matched, address
    )
    SELECT 
        sensor_id_matched,
        address,
        district AS 잘못_추출된_자치구
    FROM sensor_districts
    WHERE address IS NULL 
       OR district NOT LIKE '%구'
"""

df_anomaly = duckdb.sql(query_anomaly_check).pl()
print(f"🚨 [이상치 확인: 형식이 어긋나거나 누락된 데이터 총 {len(df_anomaly)}건]")
display(df_anomaly)
```

## 가정: 매핑되지 않은 25개 센서는 구형 스키마에 기록된 센서일 것이다
- 가정: 위경도정보 메타데이터 문서에 공시되지 않은 시리얼일 것이다.
- 결과: 꼭 그렇지는 않음 + 위치 정보 매핑이 잘 안 된 경우

```python
query_old_schema_check = """
    WITH anomaly_sensors AS (
        SELECT 
            sensor_id_matched,
            address,
            SPLIT_PART(address, ' ', 2) AS district
        FROM v_sdot_final
        GROUP BY sensor_id_matched, address
        HAVING address IS NULL OR SPLIT_PART(address, ' ', 2) NOT LIKE '%구'
    )
    SELECT 
        a.sensor_id_matched,
        a.address,
        MIN(v.measure_timestamp) AS 최초_수집일,
        MAX(v.measure_timestamp) AS 최종_수집일,
        DATE_DIFF('day', MAX(v.measure_timestamp), CAST('2026-07-19' AS TIMESTAMP)) AS 수집중단_경과일
    FROM anomaly_sensors a
    JOIN v_sdot_final v ON a.sensor_id_matched = v.sensor_id_matched
    GROUP BY a.sensor_id_matched, a.address
    ORDER BY 최종_수집일 ASC
"""

df_old_schema_check = duckdb.sql(query_old_schema_check).pl()

print("📌 [이상치 25건 센서의 수집 기간 확인]")
display(df_old_schema_check)
```

### 위치 정보 다시 매핑

```python
query_patch_view = """
    CREATE OR REPLACE VIEW v_sdot_final AS
    WITH mapped_data AS (
        SELECT 
            *,
            COALESCE(
                (SELECT 시리얼 FROM loc_table WHERE "변경 전 시리얼" = d.sensor_id OR "변경 전 시리얼(데이터 상 표기)" = d.sensor_id LIMIT 1), 
                sensor_id
            ) AS sensor_id_matched
        FROM v_sdot_clean d
    )
    SELECT 
        m.*,
        l.위도 AS lat,
        l.경도 AS lon,
        
        -- 1. 주소 텍스트 오타 및 예외 정제
        CASE 
            WHEN l.주소 LIKE '%  %' THEN REPLACE(l.주소, '  ', ' ')
            WHEN l.주소 = '서울대공원' THEN '경기도 과천시 서울대공원'
            ELSE l.주소 
        END AS address,
        
        -- 2. 공간 EDA를 위한 자치구(district) 파생 컬럼 아예 생성해두기
        CASE 
            WHEN l.주소 = '서울대공원' THEN '과천시'
            WHEN l.주소 IS NULL THEN '위치미상'
            ELSE SPLIT_PART(REPLACE(l.주소, '  ', ' '), ' ', 2)
        END AS district
        
    FROM mapped_data m
    LEFT JOIN loc_table l ON m.sensor_id_matched = l.시리얼
"""

# 뷰 업데이트 실행
duckdb.sql(query_patch_view)
print("✅ 주소 오타 및 예외 처리(서울대공원)가 반영된 v_sdot_final 뷰 업데이트 완료!\n")

# 검증 1: 보정 후 자치구별 센서 개수 다시 확인
query_verify_district = """
    SELECT 
        district AS 자치구,
        COUNT(DISTINCT sensor_id_matched) AS 고유_센서_수
    FROM v_sdot_final
    GROUP BY district
    ORDER BY 고유_센서_수 DESC
"""

df_verified = duckdb.sql(query_verify_district).pl()
print("📌 [검증 1: 보정 후 자치구별 센서 분포]")
display(df_verified)

print("-" * 50)

# 검증 2: 매핑 정합성 꼼꼼하게 최종 확인 (잔여 이상치 점검)
query_final_anomaly_check = """
    SELECT DISTINCT
        sensor_id_matched,
        address,
        district
    FROM v_sdot_final
    WHERE district NOT LIKE '%구' 
      AND district NOT IN ('과천시', '위치미상')
"""

df_final_anomaly = duckdb.sql(query_final_anomaly_check).pl()
print(f"🚨 [검증 2: 매핑 실패 잔여 이상치 총 {len(df_final_anomaly)}건]")

if len(df_final_anomaly) > 0:
    display(df_final_anomaly)
else:
    print("✅ 모든 데이터가 정상적인 '구', '과천시', 또는 '위치미상'으로 매핑되었습니다.")
```

# Export to Parquet

```python
import os

# --------------------------------------------------
# [3] v_sdot_final 뷰를 Parquet 파일로 추출 
# --------------------------------------------------
# 1. 저장할 경로 설정 (notebooks 폴더 기준 상위 디렉토리의 data/processed)
BASE_DIR = os.path.abspath('..')
SAVE_DIR = os.path.join(BASE_DIR, 'data', 'processed')

# 폴더가 없다면 안전하게 생성
os.makedirs(SAVE_DIR, exist_ok=True) 

# 최종 저장될 파일의 절대 경로
PARQUET_PATH = os.path.join(SAVE_DIR, 'sdot_final_mapped.parquet')

print(f"📂 저장 경로: {PARQUET_PATH}")
print("⏳ Parquet 파일로 추출 중입니다. 3,400만 건의 데이터라 약간의 시간이 소요될 수 있습니다...")

# 2. COPY 명령어를 사용하여 뷰의 결과를 Parquet 파일로 저장
query_export = f"""
    COPY (
        SELECT * FROM v_sdot_final
    ) TO '{PARQUET_PATH}' (FORMAT PARQUET);
"""
duckdb.sql(query_export)

print("✅ Parquet 파일 추출 완료!")
```
