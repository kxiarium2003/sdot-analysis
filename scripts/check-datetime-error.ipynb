# 시간 관련 컬럼 파싱 오류 확인 및 해결

```python
import pandas as pd
from pathlib import Path
import random

# 원본 데이터 폴더 경로 설정 (실제 경로에 맞게 수정 필요)
raw_dir = Path('../data/raw') 
csv_files = list(raw_dir.rglob('*.csv'))

# 무작위로 20개 파일 샘플링
sample_files = random.sample(csv_files, min(20, len(csv_files)))

for file_path in sample_files:
    print(f"\n📄 파일명: {file_path.name}")
    try:
        # 메모리 효율을 위해 첫 10줄만 읽기
        df = pd.read_csv(file_path, encoding='cp949', nrows=10)
        
        # S-DoT 주요 시간 관련 컬럼 찾기
        time_cols = [col for col in df.columns if col in ['전송시간', '등록일자', '측정시간']]
        
        if not time_cols:
            print("  -> ⚠️ 시간 관련 컬럼이 존재하지 않습니다.")
            continue
            
        for col in time_cols:
            sample_values = df[col].dropna().head(3).tolist()
            print(f"  - [{col}] 샘플 데이터: {sample_values}")
            
    except Exception as e:
        print(f"  -> 🚨 파일 읽기 실패: {e}")
        
print("\n✅ 진단 완료!")
```
